```python
import pandas as pd
import os

# Get the current directory (same folder as the script is run from)
mapp = os.getcwd()

# File names and corresponding years
filer_och_ar = {
    "resultat_ansokningsomgang_program_2020.xlsx": 2020,
    "resultat_ansokningsomgang_program_2021.xlsx": 2021,
    "resultat_ansokningsomgang_program_2022.xlsx": 2022,
    "resultat_ansokningsomgang_program_2023.xlsx": 2023,
    "resultat_ansokningsomgang_program_2024.xlsx": 2024
}

alla_data = []

# Read "Tabell 4" (sheet index 3) from each file
for filnamn, ar in filer_och_ar.items():
    full_sokvag = os.path.join(mapp, filnamn)
    df = pd.read_excel(full_sokvag, sheet_name="Tabell 3", engine="openpyxl")
    print(f"\n📄 {filnamn} innehåller kolumner:\n{df.columns.tolist()}\n")
    df["År"] = ar
    alla_data.append(df)

# Merge all years
sammanfogat_df = pd.concat(alla_data, ignore_index=True)

# Save result to the same folder
output_fil = os.path.join(mapp, "resultat_ansokning_program_2020-2024.xlsx")
sammanfogat_df.to_excel(output_fil, index=False)

print(f"✅ Fil sparad: {output_fil}")


import pandas as pd
import os

# Sökväg till aktuell mapp
mapp = os.getcwd()

# Filnamn och vilket år varje fil tillhör
filer_och_ar = {
    "resultat-2020-for-kurser-inom-yh-april.xlsx": 2020,
    "resultat-2020-for-kurser-inom-yh-juli.xlsx": 2020,
    "resultat-2021-for-kurser-inom-yh.xlsx": 2021,
    "resultat-2022-for-kurser-inom-yh.xlsx": 2022,
    "resultat-2023-for-kurser-inom-yh.xlsx": 2023,
    "resultat-2024-for-kurser-inom-yh.xlsx": 2024,
}

# Kolumner du vill behålla
kolumner = [
    "Beslut",
    "Anordnare namn",
    "Utbildningsområde",
    "Totalt antal beviljade platser",
    "Kommun",
    "Län"
]

alla_data = []

for filnamn, ar in filer_och_ar.items():
    full_sokvag = os.path.join(mapp, filnamn)
    df = pd.read_excel(full_sokvag, sheet_name=0)
    
    # Kontrollera om alla kolumner finns
    saknade = [col for col in kolumner if col not in df.columns]
    if saknade:
        print(f"⚠️ {filnamn} saknar kolumner: {saknade}")
        continue
    
    df = df[kolumner].copy()
    df["År"] = ar
    print(f"✅ Läste {filnamn} med {len(df)} rader.")
    alla_data.append(df)

# Slå ihop alla
sammanfogad_df = pd.concat(alla_data, ignore_index=True)

# Spara resultat
output_fil = os.path.join(mapp, "sammanfogad_kurser_inom_yh_2020-2024.xlsx")
sammanfogad_df.to_excel(output_fil, index=False)

print(f"\n✅ Färdig! Fil sparad: {output_fil}")


# import pandas as pd
# import os

# # Sökväg till aktuell mapp
# mapp = os.getcwd()

# # Filnamn och år
# filer_och_ar = {
#     "resultat-2020-for-kurser-inom-yh-april.xlsx": 2020,
#     "resultat-2020-for-kurser-inom-yh-juli.xlsx": 2020,
#     "resultat-2021-for-kurser-inom-yh.xlsx": 2021,
#     "resultat-2022-for-kurser-inom-yh.xlsx": 2022,
#     "resultat-2023-for-kurser-inom-yh.xlsx": 2023,
#     "resultat-2024-for-kurser-inom-yh.xlsx": 2024,
# }

# # Kolumner att behålla
# kolumner = [
#     "Beslut",
#     "Anordnare namn",
#     "Utbildningsområde",
#     "Totalt antal beviljade platser",
#     "Kommun",
#     "Län"
# ]

# alla_data = []

# for filnamn, ar in filer_och_ar.items():
#     full_sokvag = os.path.join(mapp, filnamn)
#     df = pd.read_excel(full_sokvag, sheet_name=0)

#     # Lägg till eventuella saknade kolumner med NaN
#     for kolumn in kolumner:
#         if kolumn not in df.columns:
#             df[kolumn] = pd.NA

#     # Behåll endast önskade kolumner (i rätt ordning)
#     df = df[kolumner].copy()
#     df["År"] = ar

#     print(f"✅ Läste {filnamn} med {len(df)} rader.")
#     alla_data.append(df)

# # Slå ihop alla
# sammanfogad_df = pd.concat(alla_data, ignore_index=True)

# # Spara till Excel
# output_fil = os.path.join(mapp, "sammanfogad_kurser_inom_yh_2020-2024.xlsx")
# sammanfogad_df.to_excel(output_fil, index=False)

# print(f"\n✅ Färdig! Fil sparad: {output_fil}")



```

```python
import pandas as pd
import os

läs_mapp = os.getcwd()
spara_mapp = os.path.abspath(os.path.join(läs_mapp, "..", "data"))

filer_och_ar = {
    "resultat-2021-for-kurser-inom-yh.xlsx": 2021,
    "resultat-2022-for-kurser-inom-yh.xlsx": 2022,
    "resultat-2023-for-kurser-inom-yh.xlsx": 2023,
    "resultat-2024-for-kurser-inom-yh.xlsx": 2024,
}

kolumner = [
    "Beslut",
    "Anordnare namn",
    "Utbildningsområde",
    "Totalt antal beviljade platser",
    "Kommun",
    "Län"
]

alla_data = []

for filnamn, ar in filer_och_ar.items():
    full_sokvag = os.path.join(läs_mapp, filnamn)
    df = pd.read_excel(full_sokvag, sheet_name=0)
    df.columns = df.columns.str.strip()  # Viktigt! Tar bort mellanslag runt kolumnnamn

    for kolumn in kolumner:
        if kolumn not in df.columns:
            df[kolumn] = pd.NA

    df = df[kolumner].copy()
    df["År"] = ar
    print(f"✅ Läste {filnamn} med {len(df)} rader.")
    alla_data.append(df)

sammanfogad_df = pd.concat(alla_data, ignore_index=True)
output_kurser = os.path.join(spara_mapp, "sammanfogad_kurser_2020-2024.xlsx")
sammanfogad_df.to_excel(output_kurser, index=False)
print(f"💾 Sparad till: {output_kurser}")



```
